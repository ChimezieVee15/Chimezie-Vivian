### 4.2 Variables: Storing Information

Variables are like labeled boxes that store information. In linguistics, we might store words, sentences, or analysis results.

```python
# Creating variables
wordO1 = "linguistics"
sentence = "I love Ibadan"
language_count = 42

# Using variables
print('Testing')
print(wordO1)
print(sentence)
print("Number of languages in our database:")
print(language_count)
```

**Variable Rules:**
- Names can contain letters, numbers, and underscores
- Must start with a letter or underscore (not a number)
- Case-sensitive: `Word` and `word` are different
- Use descriptive names: `sentence` is better than `s`

```python
# Good variable names for
source_text = "The quick brown fox jumps over the lazy dog."
word_count = 9
is_english = True
phoneme_inventory = ["p", "t", "k", "b", "d", "g"]

print(F"Text: {source_text}")
print(f"Word count: {word_count}")
print(f"Is this English?: {is_english}")
```

### 5.2 Numbers (Integers and Floats)

Numbers are essential for linguistic analysis - counting, measuring, calculating frequencies.

```python
# Integers (whole numbers)
word_count = '150'
word_count2 = 150
sentence_count = 12
phoneme_count = 44  # English phonemes

# Floats (decimal numbers)
average_word_length = 4.7
frequency_ratio = 0.023
similarity_score = 0.85
print(type('frequency_ratio'))
print(type(word_count))
# Basic arithmetic
total_words = sentence_count + 50
total_words2 = word_count + " no matter"
words_per_sentence = word_count2 / sentence_count
print(total_words)
print(total_words2)
print(f"\nCalculations:")
print(f"Total words after adding 50: {total_words}")
print(f"Average words per sentence: {words_per_sentence:.5f}")
```

### 5.3 Booleans (True/False Values)

Booleans represent yes/no, true/false conditions - very useful for linguistic analysis.

```python
# Boolean values
is_verb = True
is_plural = False
has_prefix = True
is_english_word = False

#print("Boolean examples:")
#print(f"Word is a verb: {is_verb}")
#print(f"Word is plural: {is_plural}")
#print(f"Word has prefix: {has_prefix}")
#print(f"Word is English: {is_english_word}")

# Booleans from comparisons
word = "No way"
word_length = len(word)  # len() gives us the length

is_long_word = word_length > 6
starts_with_r = word.startswith('r')
ends_with_ing = word.endswith('ing')

print(f"\nAnalyzing the word '{word}':")
print(f"Length: {word_length} characters")
print(f"Is long word (>6 chars): {is_long_word}")
print(f"Starts with 'r': {starts_with_r}")
print(f"Ends with 'ing': {ends_with_ing}")
```

### 5.4 Checking Data Types

You can check what type of data you're working with using the `type()` function.

```python
# Examples of each data type
text_data = "morphology"
count_data = 25
decimal_data = 3.14
boolean_data = True

print("Data type examples:")
print(f"'{text_data}' is type: {type(text_data)}")
print(f"{count_data} is type: {type(count_data)}")
print(f"{decimal_data} is type: {type(decimal_data)}")
print(f"{boolean_data} is type: {type(boolean_data)}")

# This is useful for debugging!
mystery_variable = "42"
print(f"\nWhat type is '42'? {type(mystery_variable)}")
print("Notice: '42' in quotes is a string, not a number!")
```
