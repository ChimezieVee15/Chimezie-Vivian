### 7.2 Morphological Analysis Practice

```python
# Simple morphological analysis
def analyze_word(word):
    """Basic morphological analysis of English words"""
    word = word.lower().strip()
    analysis = {'word': word, 'length': len(word)}
    
    # Check for common prefixes
    prefixes = ['un-', 're-', 'pre-', 'dis-', 'in-', 'im-']
    for prefix in prefixes:
        if word.startswith(prefix[:-1]):  # Remove the dash
            analysis['prefix'] = prefix[:-1]
            break
    
    # Check for common suffixes
    suffixes = ['-ing', '-ed', '-er', '-est', '-ly', '-tion', '-ness']
    for suffix in suffixes:
        if word.endswith(suffix[1:]):  # Remove the dash
            analysis['suffix'] = suffix[1:]
            break
    
    return analysis

# Test words
test_words = ['unhappy', 'running', 'teacher', 'beautiful', 'reconstruction', 'quickly']

print("MORPHOLOGICAL ANALYSIS")
print("=" * 40)

for word in test_words:
    analysis = analyze_word(word)
    print(f"\nWord: '{analysis['word']}'")
    print(f"Length: {analysis['length']} characters")
    
    if 'prefix' in analysis:
        print(f"Prefix: {analysis['prefix']}")
    
    if 'suffix' in analysis:
        print(f"Suffix: {analysis['suffix']}")
    
    # Extract potential root
    root = word
    if 'prefix' in analysis:
        root = root[len(analysis['prefix']):]
    if 'suffix' in analysis:
        root = root[:-len(analysis['suffix'])]
    
    if root != word:
        print(f"Potential root: '{root}'")
```

### 7.3 Phonetic Analysis (Basic)

```python
# Basic phonetic analysis
def count_syllables(word):
    """Count the number of syllables in an English word.
    This is an approximation based on vowel groups."""
    word = word.lower().strip(".:;?!")
    
    # Handle empty words
    if not word:
        return 0
    
    # Count vowel groups
    vowel_groups = 0
    in_vowel_group = False
    
    for char in word:
        if char in 'aeiouy':
            if not in_vowel_group:  # Start of a new vowel group
                vowel_groups += 1
                in_vowel_group = True
        else:
            in_vowel_group = False
    
    # Handle special cases
    # Words ending with 'e' often don't form a separate syllable
    if word.endswith('e'):
        vowel_groups -= 1
    
    # Words ending with 'le' after a consonant often form a syllable
    if word.endswith('le') and len(word) > 2 and word[-3] not in 'aeiouy':
        vowel_groups += 1
    
    # Every word has at least one syllable
    return max(1, vowel_groups)

def phonetic_analysis(word):
    """Simple phonetic analysis of English words"""
    word = word.lower()
    
    # Define vowels and consonants
    vowels = 'aeiou'
    consonants = 'bcdfghjklmnpqrstvwxyz'
    
    # Count phonetic features
    vowel_count = sum(1 for char in word if char in vowels)
    consonant_count = sum(1 for char in word if char in consonants)
    
    # Find vowel and consonant positions
    vowel_positions = [i for i, char in enumerate(word) if char in vowels]
    consonant_positions = [i for i, char in enumerate(word) if char in consonants]
    
    # Get syllable count
    syllable_count = count_syllables(word)
    
    # Get CV structure
    structure = ""
    for char in word:
        if char in 'aeiou':
            structure += 'V'
        else:
            structure += 'C'
    
    return {
        'word': word,
        'vowels': vowel_count,
        'consonants': consonant_count,
        'vowel_positions': vowel_positions,
        'syllable_estimate': syllable_count,
        'cv_structure': structure
    }

# Test words
phonetic_test_words = ['cat', 'beautiful', 'strength', 'linguistics', 'amazing', 'apple', 'bottle']

print("PHONETIC ANALYSIS")
print("=" * 40)

for word in phonetic_test_words:
    analysis = phonetic_analysis(word)
    
    print(f"\nWord: '{analysis['word']}'")
    print(f"Vowels: {analysis['vowels']}")
    print(f"Consonants: {analysis['consonants']}")
    print(f"Estimated syllables: {analysis['syllable_estimate']}")
    print(f"CV structure: {analysis['cv_structure']}")
```

### Exercise 1: Word Length Analysis

```python
# Exercise 1: Analyze the lengths of words in this sentence
sentence = "Computational linguistics combines computer science and linguistic theory"

# Your task: 
# 1. Split the sentence into words
# 2. Find the longest and shortest words
# 3. Calculate the average word length
# 4. Count how many words are longer than 6 characters

# Try writing your solution here:

# SOLUTION:
def analyze_word_lengths(sentence: str) -> dict:
    """Analyze word lengths in a sentence.
    
    Args:
        sentence: Input string to analyze
        
    Returns:
        dict: Analysis results including word counts, lengths, and statistics
    """
    if not sentence.strip():
        return {"error": "Empty input string"}
        
    # Split and clean words (remove punctuation)
    words = [word.strip(".,!?") for word in sentence.split()]
    
    if not words:
        return {"error": "No valid words found"}
    
    word_lengths = [len(word) for word in words]
    
    return {
        "word_count": len(words),
        "longest_word": max(words, key=len),
        "longest_length": max(word_lengths, default=0),
        "shortest_word": min(words, key=len),
        "shortest_length": min(word_lengths, default=0),
        "average_length": round(sum(word_lengths) / len(word_lengths), 2),
        "long_words": [word for word in words if len(word) > 6],
        "long_word_count": sum(1 for word in words if len(word) > 6)
    }

def print_analysis(sentence: str) -> None:
    """Print formatted word length analysis."""
    print("=" * 50)
    print("WORD LENGTH ANALYSIS".center(50))
    print("=" * 50)
    
    result = analyze_word_lengths(sentence)
    
    if "error" in result:
        print(f"Error: {result['error']}")
        return
        
    print(f"\n📝 Sentence: {sentence}")
    print(f"🔢 Number of words: {result['word_count']}")
    print(f"\n📏 Word Lengths:")
    print(f"   Longest: '{result['longest_word']}' ({result['longest_length']} chars)")
    print(f"   Shortest: '{result['shortest_word']}' ({result['shortest_length']} chars)")
    print(f"   Average: {result['average_length']} characters")
    
    print(f"\n📊 Words > 6 characters:")
    print(f"   Count: {result['long_word_count']}")
    if result['long_word_count'] > 0:
        print(f"   Words: {', '.join(result['long_words'])}")
    print("=" * 50 + "\n")

# Example usage
if __name__ == "__main__":
    test_sentence = "Computational linguistics combines computer science and linguistic theory"
    print_analysis(test_sentence)
    
    # Test edge cases
    print_analysis("")  # Empty string
    print_analysis("...")  # Only punctuation
    print_analysis("a be cat")  # Short words
```
