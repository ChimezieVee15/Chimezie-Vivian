```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def inorder(node):
    if node:
        inorder(node.left) # 1. Go left
        print(node.value, end=" ") # 2. Visit the node
        inorder(node.right) # 3. Go right

def preorder(node):
    if node:
        print(node.value, end=" ") # 1. Visit the node
        preorder(node.left) # 2. Go left  
        preorder(node.right) # 3. Go right

def postorder(node):
    if node:
        postorder(node.left) # 1. Go left  
        postorder(node.right) # 2. Go right
        print(node.value, end=" ") # 3. Visit the node

# Creating a sample binary tree
root = Node("A") # Root node
#chiildren of A
root.left = Node("B") # Left child of root
root.right = Node("C") # Right child of root

#children of B
root.left.left = Node("D") # Left child of B 
root.left.right = Node("E") # Right child of B
#children of C
root.right.left = Node("F") # Left child of C 
root.right.right = Node("G") # Right child of 

print("Inorder Traversal:")
inorder(root) 
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# Traversals
def inorder(node):
    if node:
        inorder(node.left)
        print(node.value, end=" ")
        inorder(node.right)

def preorder(node):
    if node:
        print(node.value, end=" ")
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.value, end=" ")

# Build semi-complex tree
root = Node("M") # Root node 
#chiildren of M
root.left = Node("C")
root.right = Node("R")
#children of C

root.left.left = Node("A")
root.left.right = Node("F")
#children of R

root.right.left = Node("P")
root.right.right = Node("S")
#children of F

root.left.right.left = Node("E")

# Print traversals
print("Inorder:")
inorder(root)

print("\nPreorder:")
preorder(root)

print("\nPostorder:")
postorder(root)
```

```python
# Step 1: Create a dictionary
person = {"name": "Vivian", "age": 21, "city": "Lagos"}
print("Step 1: Original Dictionary:", person)

# Step 2: Access items
print("Step 2: Access Name:", person["name"])
print("Step 2: Access City:", person["city"])

# Step 3: Add a new key-value pair
person["hobby"] = "Coding"
print("Step 3: After Adding Hobby:", person)

# Step 4: Change a value
person["city"] = "Abuja"
print("Step 4: After Changing City:", person)

# Step 5: Remove a key-value pair
del person["age"]
print("Step 5: After Removing Age:", person)

# Step 6: Loop through dictionary
print("Step 6: Loop through dictionary:")
for key, value in person.items():
    print(key, ":", value)
```
