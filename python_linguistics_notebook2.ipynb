## 6. Basic String Operations for Linguistic Data

Strings are the heart of linguistic programming. Let's explore the most important operations.

### 6.1 String Length and Basic Properties

```python
# Working with linguistic text
sentence = "The quick brown fox jumps over the lazy dog."
word = "internationalization"
phonetic = "ɪntərnæʃənəlaɪzeɪʃən"

print("String length analysis:")
print(f"Sentence: '{sentence}'")
print(f"Length: {len(sentence)} characters")
print(f"Word: '{word}'")
print(f"Length: {len(word)} characters")
print(f"Phonetic: '{phonetic}'")
print(f"Length: {len(phonetic)} characters")

```

```python
# Character vs. word length comparison
words_in_sentence = sentence.split()  # Split into words
print(f"\nSentence has {len(sentence)} characters but {len(words_in_sentence)} words")
```

### 6.2 Accessing Parts of Strings (Indexing and Slicing)

You can access individual characters or portions of strings - very useful for morphological analysis!

```python
word = "linguistics"
print(f"Working with: '{word}'")
print(f"Length: {len(word)}")
```

```python
# Indexing (accessing individual characters)
# Python uses 0-based indexing!
print("\nCharacter access:")
print(f"First character (index 0): '{word[0]}'")
print(f"Second character (index 1): '{word[1]}'")
print(f"Last character (index -1): '{word[-1]}'")
print(f"Second to last (index -2): '{word[-2]}'")
```

```python
# Slicing (accessing parts of strings)
print("\nString slicing:")
print(f"First 4 characters: '{word[0:4]}'")
print(f"From index 2 to 6: '{word[2:6]}'")
print(f"Last 4 characters: '{word[-4:]}'")
print(f"Everything except last 2: '{word[:-2]}'")
```

```python
# Linguistic applications
print("\nLinguistic analysis:")
prefix = word[:4]  # "ling"
suffix = word[-3:]  # "ics"
root = word[4:-3]  # "uist"

print(f"Potential prefix: '{prefix}'")
print(f"Potential suffix: '{suffix}'")
print(f"Potential root: '{root}'")
```

### 6.4 Finding and Checking Content

Essential for searching through linguistic data.

```python
text = "The phoneme /θ/ appears in think and thought."
print(f"Text: '{text}'")

# Searching for substrings
print("\nSearching for content:")
print(f"Contains 'phoneme': {'phoneme' in text}")
print(f"Contains 'morpheme': {'morpheme' in text}")
print(f"Position of 'think': {text.find('think')}")
print(f"Position of 'xyz': {text.find('xyz')}")  # Returns -1 if not found
```

```python
# String checking methods
word1 = "running"
word2 = "cat"
word3 = "123"

print("\nString analysis:")
print(f"'{word1}' starts with 'run': {word1.startswith('run')}")
print(f"'{word1}' ends with 'ing': {word1.endswith('ing')}")
print(f"'{word2}' is alphabetic: {word2.isalpha()}")
print(f"'{word3}' is numeric: {word3.isdigit()}")
print(f"'{word2}' is alphabetic: {word2.isdigit()}")
```

```python
# Counting occurrences
sentence = "The cat in the hat sat on the mat with the bat."
print(f"\nCounting in: '{sentence}'")
print(f"'the' appears {sentence.lower().count('the')} times")
print(f"'at' appears {sentence.count('at')} times")
```

### 6.5 Splitting and Joining Strings

Crucial for breaking text into words and reconstructing it.

```python
# Splitting strings into lists
sentence = "Natural language processing is amazing"
print(f"Original sentence: '{sentence}'")
```

```python
# Split by spaces (default)
words = sentence.split()
print(f"Words: {words}")
print(f"Number of words: {len(words)}")
```

```python
# Split by specific characters
csv_data = "word,frequency,pos,length"
fields = csv_data.split(',')
print(f"\nCSV fields: {fields}")
```

```python
# Linguistic example: morpheme boundaries
morphemes = "un-believe-able"
parts = morphemes.split('-')
print(f"Morphemes in '{morphemes}': {parts}")
```

```python
# Joining strings back together
word_list = ['computational', 'linguistics', 'rocks']
joined_sentence = ' '.join(word_list)
print(f"\nJoining words: {joined_sentence}")
```

```python
# Join with different separators
phonemes = ['k', 'æ', 't']
ipa_notation = '.'.join(phonemes)
print(f"Phonemes: {phonemes}")
print(f"IPA notation: [{ipa_notation}]")
```

```python
# Practical example: word frequency analysis preparation
text = "The quick brown fox jumps over the lazy dog. The dog was sleeping."
words = text.lower().replace('.', '').split()
print(f"\nText processing example:")
print(f"Original: '{text}'")
print(f"Processed words: {words}")
print(f"Unique words: {set(words)}")  # set() removes duplicates
```

### 6.6 String Replacement and Formatting

Useful for data cleaning and text normalization.

```python
# String replacement
original = "I luv programming and coding and development"
print(f"Original: '{original}'")

# Simple replacement
corrected = original.replace('luv', 'love')
print(f"Corrected: '{corrected}'")
```

```python
# Multiple replacements for normalization
text_with_variants = "colour and colour and color and colors"
normalized = text_with_variants.replace('colour', 'color')
print(f"\nNormalization example:")
print(f"Original: '{text_with_variants}'")
print(f"Normalized: '{normalized}'")
```

```python
# Removing punctuation
sentence_with_punct = "Hello, world! How are you?"
no_punct = sentence_with_punct.replace(',', '').replace('!', '').replace('?', '')
print(f"\nPunctuation removal:")
print(f"With punctuation: '{sentence_with_punct}'")
print(f"Without punctuation: '{no_punct}'")
```

```python
# String formatting for reports
word = "fascinating"
length = len(word)
vowels = sum(1 for char in word.lower() if char in 'aeiou')
consonants = length - vowels

print(f"\nLinguistic analysis report:")
print(f"Word: {word}")
print(f"Total length: {length} characters")
print(f"Vowels: {vowels}")
print(f"Consonants: {consonants}")
print(f"Vowel ratio: {vowels/length:.2%}")
```
